```python
import pandas as pd
import plotly.graph_objects as go

# Data
events = [
    ("Launch of iPhone", 17),
    ("Founding of USA", 249),
    ("Fall of Roman Empire", 1550),
    ("Construction of Great Pyramid", 4500),
    ("End of Last Ice Age", 11700),
    ("Spent nuclear fuel", 200000)
]

df = pd.DataFrame(events, columns=["Event", "Years Ago"])

# Figure
fig = go.Figure()

# Bar trace (blue theme)
fig.add_trace(go.Bar(
    x=df["Event"],
    y=df["Years Ago"],
    marker_color="#81c500",  # sleek deep blue
    hovertemplate="<b>%{x}</b><br>%{y:,} years ago",
    showlegend=False
))

# Horizontal line for nuclear fuel
fig.add_hline(
    y=200000,
    line_dash="dash",
    line_color="#d62728", # refined red
    annotation_text="☢️ Spent nuclear fuel remains radioactive for ~200,000 years",
    annotation_position="top right",
    annotation_font_color="#d62728",
    annotation_font_size=13
)

# Layout
fig.update_layout(
    title=dict(
        text="<b>Perspective of Time:</b> Comparing Human History to Nuclear Waste Longevity",
        x=0.5,
        font=dict(size=22, color="#0b0c10", family="Arial")
    ),
    xaxis=dict(
        title="<b>Key Events in Human History</b>",
        tickangle=-35,
        tickfont=dict(size=13, color="#0b0c10"),
        showgrid=False,
        linecolor="#cccccc"
    ),
    yaxis=dict(
        title="<b>Years Ago</b>",
        tickfont=dict(size=13, color="#0b0c10"),
        showgrid=True,
        gridcolor="#e5e5e5",
        linecolor="#cccccc"
    ),
    plot_bgcolor="#ffffff",
    paper_bgcolor="#ffffff",
    width=900,
    height=550,
    margin=dict(l=80, r=80, t=90, b=120),
)

# Optional: subtle hover effect color consistency
fig.update_traces(marker_line_width=0.5, marker_line_color="#0b0c10")

fig.show()

```

```python
import pandas as pd
import plotly.graph_objects as go

# Data
data = [
    ("African Bush Elephant", 6),
    ("Semi-Truck + Trailer", 36),
    ("Fighter Jet", 23),
    ("Boeing 747", 400),
    ("U.S. Annual Spent Nuclear Fuel", 2000)
]

df = pd.DataFrame(data, columns=["Item", "Metric Tons"])

# Figure
fig = go.Figure()

# Bar trace with uniform royal blue
fig.add_trace(go.Bar(
    x=df["Item"],
    y=df["Metric Tons"],
    text=[f"{y:,} tons" for y in df["Metric Tons"]],
    textposition="outside",
    marker_color="#81c500",  # royal blue
    hovertemplate="<b>%{x}</b><br><b>Weight:</b> %{y:,} metric tons<extra></extra>"
))

# Annotation for nuclear fuel
fig.add_annotation(
    x="U.S. Annual Spent Nuclear Fuel",
    y=2000,
    text="≈ 333 elephants • ≈ 5 Boeing 747s",
    showarrow=True,
    arrowhead=2,
    arrowcolor="red",
    font=dict(size=13, color="red"),
    ay=-50
)

# Layout styling
fig.update_layout(
    title=dict(
        text="<b>How Heavy Is Spent Nuclear Fuel?</b><br><sup>Comparing radioactive waste to familiar massive objects</sup>",
        x=0.5,
        font=dict(size=22, color="#0b0c10", family="Arial")
    ),
    xaxis=dict(
        title="",
        tickangle=-15,
        tickfont=dict(size=14, color="#0b0c10"),
        showgrid=False,
        linecolor="#cccccc"
    ),
    yaxis=dict(
        title="<b>Weight (Metric Tons)</b>",
        tickfont=dict(size=13, color="#0b0c10"),
        showgrid=True,
        gridcolor="#e5e5e5",
        linecolor="#cccccc"
    ),
    plot_bgcolor="#ffffff",
    paper_bgcolor="#ffffff",
    width=950,
    height=600,
    margin=dict(l=80, r=60, t=100, b=100),
    font=dict(size=14),
)

# Slight outline for bar definition
fig.update_traces(marker_line_width=0.8, marker_line_color="#0b0c10")

fig.show()

```

```python
import pandas as pd
import plotly.express as px

# 1. Load the Excel data
df = pd.read_excel("summary3.xlsx")

# 2. Select relevant columns
df = df[[
    "Year",
    "BWR Number of assemblies in Pool",
    "PWR Number of assemblies in Pool",
    "Total Number of assemblies in Pool"
]]

# 3. Rename columns
df = df.rename(columns={
    "BWR Number of assemblies in Pool": "BWR",
    "PWR Number of assemblies in Pool": "PWR",
    "Total Number of assemblies in Pool": "Total"
})

# 4. Melt to long format
df_long = df.melt(
    id_vars=["Year", "Total"],
    value_vars=["BWR", "PWR"],
    var_name="Reactor Type",
    value_name="Assemblies"
)

# 5. Filter to show every 5 years
df_long = df_long[df_long["Year"] % 5 == 0]

# 6. Create grouped bar chart
fig = px.bar(
    df_long,
    x="Year",
    y="Assemblies",
    color="Reactor Type",
    hover_data={"Total": True},
    title="<b>Spent Nuclear Fuel (BWR & PWR) in U.S. Pools (1970–2020)</b>",
    barmode="group",
    color_discrete_map={
        "BWR": "#81c500",   # Sleek royal blue
        "PWR": "#d3e424"    # Coral orange for contrast
    }
)

# 7. Layout polish for publication
fig.update_layout(
    title=dict(
        font=dict(size=22, color="#0b0c10", family="Arial"),
        x=0.5
    ),
    xaxis=dict(
        title="<b>Year</b>",
        tickfont=dict(size=13, color="#0b0c10"),
        showgrid=False,
        linecolor="#cccccc"
    ),
    yaxis=dict(
        title="<b>Number of Assemblies in Pool</b>",
        tickfont=dict(size=13, color="#0b0c10"),
        showgrid=True,
        gridcolor="#e6e6e6",
        linecolor="#cccccc"
    ),
    legend=dict(
        title="<b>Reactor Type</b>",
        orientation="v",
        x=1,
        y=1,
        bgcolor="rgba(255,255,255,0.6)",
        bordercolor="#dcdcdc",
        borderwidth=1
    ),
    font=dict(size=14),
    template="simple_white",
    plot_bgcolor="#ffffff",
    paper_bgcolor="#ffffff",
    width=1000,
    height=550,
    margin=dict(l=70, r=70, t=90, b=70)
)

# 8. Subtle bar outline for definition
fig.update_traces(marker_line_width=0.8, marker_line_color="#0b0c10")

fig.show()
```
