```python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy
site_filepath='/Users/melodyqian/Documents/GitHub/FindMyNuclearWaste/data/MLDataset.csv'
df=pd.read_csv(site_filepath)
```

```python
fig = go.Figure(data=go.Choropleth(
    locations=df['state_abbrev'], # Spatial coordinates
    locationmode = 'USA-states', # set of locations match entries in `locations`
    colorscale = 'RdYlGn',
    showscale=True,
    zmax=50000,
    zmin=-50000,
    zmid=0
))

scatter_trace  = go.Scattergeo(
        lon = df['longitude'],
        lat = df['latitude'],
        mode = 'markers',
        marker=dict(
            color='rgb(211, 228, 36)',
            size=4,
            line=dict(
                color='Black',
                width=.5
            )
        )
        )
fig.add_trace(scatter_trace)
fig.update_layout(
    title_text = 'Sites',
    geo_scope='usa', # limite map scope to USA
)

fig.show()
```
